# Régression

On souhaite à présent effectuer des regressions afin d'observer la correlation entre le niveau de richesse des régions et les indicateurs des dépenses. 
On effectue donc des regressions simples unidimentionnels des indicateurs de dépenses les plus pertnents, à savoir les dépenses des régimes d'assurance maladoe et le FIR (Fond d'intervention régional)

$$ \widehat{Y}_i = \widehat{\beta}_0 + \widehat{\beta}_1 X_i $$

Avec :
- $\widehat{Y}_i$ : les dépenses de santé estimées pour la région $i$ ;
- $X_i$ : le PIB par habitant de la région $i$ ;
- $\widehat{\beta}_0$ : la constante estimée ;
- $\widehat{\beta}_1$ : le coefficient directeur estimé, qui mesure l'association entre le PIB par habitant et les dépenses de santé.


```python
# Base de données des états financiers (année 2023)
etats_financiers = projet.generate_etats_financiers_regionaux()
df_financier = etats_financiers.copy()
display(df_financier.head())

# Harmoniser le nom de la région avec la table PIB
df_financier["Région"] = df_financier["Région"].replace({
    "Auvergne-Rhônes-Alpes": "Auvergne-Rhône-Alpes"
})
```

```python
print(df_financier)
```

A ce stade, nos deux bases de données pib_2023 (PIB par habitant en 2023) et df_financier ( états financier des régions en 2023) contiennent le même nombre de régions avec les mêmes noms. 


```python
print(df_financier.columns.tolist())
```

```python
# Transformer les catégories en colonnes
df_financier_pivot = df_financier.pivot( index="Région", columns="Catégorie", values="Montant 2023 (M€)").reset_index()

# Ajouter le PIB par habitant
df_analyse = pd.merge( df_financier_pivot, pib_2023, on="Région", how="inner" )
```

```python
print("Colonnes df_financier :", df_financier.columns.tolist())
print("Colonnes pib_2023 :", pib_2023.columns.tolist())
print("Colonnes df_analyse :", df_analyse.columns.tolist())
```

```python
# Variable explicative commune aux trois régressions
X = df_analyse[["PIB_par_habitant"]]
X = sm.add_constant(X)

variable_explicative = "PIB_par_habitant"
variables_expliquees = [
    "DEPENSES DES REGIMES D'ASSURANCE MALADIE (hors contribution aux dépenses des ESMS pour personnes âgées et handicapées)",
    "FONDS D'INTERVENTION REGIONAL (FIR)"
]
```

```python
# Réalisation des régressions

resultats = regression.plusieurs_regressions(df_analyse, variable_explicative, variables_expliquees)

for variable, modele in resultats.items():
    print("\n" + "=" * 80)
    print("VARIABLE EXPLIQUÉE :", variable)
    print("=" * 80)
    print(modele.summary())
```

# Identification des profils des établissements de santé

Dans cette partie, on souhaite identifier des profils d'établissements présentant des caractéristiques similaires. 
Nous utilisons les données de la SAE (Statistique Annuelle des Établissements de santé) de 2023 où chaque observation correspond à un établissement de santé. Cette base regroupe des informations sur les établissements de santé français, notamment leur activité, leurs capacités d'accueil et leur personnel. Nous nous intéressons plus particulièrement à la table MCO (Médecine, Chirurgie et Obstétrique), qui contient des données sur les lits disponibles, les séjours hospitaliers et les journées d'hospitalisation, etc.

À l'aide du fichier Excel qui accompagne la base, nous pouvons associer chaque nom de variable à sa signification. Cette étape nous permet de comprendre les informations disponibles et de sélectionner les variables les plus pertinentes pour notre étude. L'objectif est de retenir des indicateurs qui permettront de caractériser les établissements de santé et de les regrouper selon leurs profils grâce à une méthode de classification (K-means).

## Récupération des données et chois des variables

```python
# On récupère le dictionnaire de variables
mco = projet.extract_mco_sae()
dictionnaire = projet.extract_dictionnaire_sae()
dic = dictionnaire["SAE 2023"]

# On filtre uniquement les variables de la table MCO
dic_mco = dic[dic["Table SAS"] == "MCO"]

# On récupère les colonnes présentes dans notre CSV
colonnes_mco = pd.DataFrame({"Variable SAS": mco.columns})

# On fait la correspondance avec le dictionnaire Excel
signification_mco = colonnes_mco.merge(dic_mco[["Variable SAS", "Label variable SAS"]], how="left")
```

```python
# Signification des colonnes de la base de données
pd.set_option("display.max_colwidth", None)
display(signification_mco)
```

Nous choisissons comme variables à notre clstering : 
    - "LIT_MCO" : Nombre de lits installés en hospitalisation complète
    - "JLI_MCO" : Nombre de journées exploitables de MCO
    - "SEJHC_MCO" : Nombre de séjours en hospitalisation complète
    - "JOU_MCO" : Nombre de journées d'hospitalisation
    - "PLA_MCO" : Nombre de places en hospitalisation partielle ou ambulatoire
    - "SEJHP_MCO" : Nombre de séjours en hospitalisation partielle ou ambulatoire
    - "SEA_MCO" : Nombre de séances de MCO
Ces variables permettent de prendre en compte plusieurs modalités de prise en charge : l'hospitalisation complète, l'hospitalisation partielle ou ambulatoire et les séances.
Nous avons privilégié les indicateurs globaux du secteur MCO plutôt que les variables détaillant séparément la médecine, la chirurgie et l'obstétrique afin de rester sur une vision d'ensemble de l'activité des établissements.

```python
var = ["LIT_MCO", "JLI_MCO", "SEJHC_MCO", "JOU_MCO", "PLA_MCO", "SEJHP_MCO", "SEA_MCO"]
```

```python
mco_selection, mco_standardise = processing.prepare_mco_data(mco, var)
#Vérification des doublons
print("Nombre de doublons :", mco.duplicated(subset="FI").sum())
```

## Détermination du nombre optimal de clusters

```python
print(len(mco_selection))
```

Pour choisir le nombre de clusters, nous utilisons la méthode du coude. Elle consiste à tester plusieurs nombres de clusters et à observer l'évolution de l'inertie pour chacun.

Nous représentons ensuite ces valeurs sur un graphique afin de repérer le point où la diminution de l'inertie devient moins importante. Ce point nous aide à choisir le nombre de clusters à utiliser pour notre analyse.

```python
#Application de la méthode kmeans :
kmeans.methode_coude(mco_standardise)
```

On choisit alors 3 comme nombre de clusters, par la méthode du coude.

## Méthode k-means

```python
clusters = kmeans.appliquer_kmeans(mco_standardise, 3)
mco_selection["Cluster"] = clusters
```

```python
print(mco_selection["Cluster"].value_counts().sort_index())
```

```python
# Comparaison des clusters
profil_clusters = mco_selection.groupby("Cluster")[
    ["LIT_MCO", "JLI_MCO", "SEJHC_MCO", "JOU_MCO",
     "PLA_MCO", "SEJHP_MCO", "SEA_MCO"]
].mean()

print(profil_clusters)
```

On observe que le cluster 0 correspond en moyenne à peu de lits et une activité relativement faible, que le cluster 1 correspond à beaucoup de lits, de journées et de séjours, il regroupe les établissements ayant en moyenne des volumes d'activité les plus importants. Le Cluster 2 regrouperait quand à lui des valeurs intermédiaires entre les deux autres groupes.

### Répartition par département

Le numéro FINESS (Fichier National des Établissements Sanitaires et Sociaux) est un identifiant unique de 9 caractères dont disposent les établissements sanitaire. En général, ses deux premiers caractères indiquent le département dans lequel il se situe. C'est ce qui va nous permettre de le localiser sur la carte des départements de France.

```python
# On récupère le FINESS et le departement
mco_selection["FI"] = mco.loc[mco_selection.index, "FI"].astype(str).str.zfill(9)
mco_selection["Departement"] = mco_selection["FI"].apply(
    lambda x: x[:3] if x.startswith("97") else x[:2]
)

#On vérifie les départements
print(sorted(mco_selection["Departement"].unique()))
```

```python
#Répartition des clusters par département
repartition_departements = mco_selection.groupby(
    ["Departement", "Cluster"]
).size().unstack(fill_value=0)

visualisation.afficher_repartition_clusters_departements(mco_selection)
```

```python
visualisation.visualiser_clusters_departements(mco_selection)
```

On constate que pour la plupart des régions ce sont les établissements de type "cluster 0" qui prédominent. Les établissements de type "2" sont plus présents dans les départements 75,

## Analyse de la répartition géographique des établissements par cluster

Le cluster 0 regroupe un nombre important d'établissements à Paris (75), dans le Nord (59) et dans les Bouches-du-Rhône (13). Toutefois, il est proportionnellement dominant dans certains départements moins représentés en nombre, comme la Creuse (23), la Haute-Marne (52) et la Lozère (48), où tous les établissements recensés dans notre échantillon appartiennent à ce cluster.

Le cluster 1 est principalement représenté en nombre à Paris (75). En proportion, il occupe une place plus importante dans le Morbihan (56), où il représente 20 % des établissements du département, suivi de Paris (75), avec environ 17 %. Il reste cependant minoritaire dans les départements étudiés.

Le cluster 2 est particulièrement présent en nombre dans le Nord (59), le Rhône (69), les Bouches-du-Rhône (13) et l'Hérault (34)

La mise en perspective avec le PIB par habitant régional fait apparaître des situations variées. Paris et les Hauts-de-Seine appartiennent à l'Île-de-France, qui présente le PIB par habitant le plus élevé de notre classement, avec environ 66 753 € en 2023. Le Rhône et les Bouches-du-Rhône se situent également dans des régions au PIB relativement élevé. À l'inverse, le Nord appartient aux Hauts-de-France, dont le PIB par habitant est d'environ 31 959 €. Les départements les plus représentés dans les clusters ne se concentrent donc pas uniquement dans les régions les plus riches.

Cette comparaison comporte plusieurs limites : le PIB est mesuré à l'échelle régionale, tandis que le clustering est étudié à l'échelle départementale. De plus, les proportions calculées peuvent être influencées par le faible nombre d'établissements dans certains départements. Enfin, notre étude porte uniquement sur les activités de médecine, chirurgie et obstétrique (MCO). Les profils identifiés ne représentent donc pas l'ensemble des activités hospitalières.
